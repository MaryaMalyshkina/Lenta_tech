```python
import torch
from PIL import Image
```

```python
from ultralytics import YOLO
```

```python
# Проверка доступности GPU
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print(f"Using device: {device}")
```

```python
import yaml
import os

# Создаём правильный data.yaml с абсолютными путями
data_config = {
    'train': '/kaggle/input/datasets/mashamalyshkina/lenta-yolo-dataset/train/images',
    'val': '/kaggle/input/datasets/mashamalyshkina/lenta-yolo-dataset/train/images',
    'nc': 1,
    'names': ['price tag']
}

# Сохраняем в рабочую директорию
with open('/kaggle/working/data.yaml', 'w') as f:
    yaml.dump(data_config, f)

print("data.yaml создан")
```

```python
model = YOLO("yolo11n.pt")        # или "yolo26n.pt"

model.train(
    data="/kaggle/working/data.yaml",
    epochs=500,           
    imgsz=640,
    batch=16,             
    patience=30,
    augment=True,
    optimizer='auto',
    device='cpu',             # 0 = GPU, 'cpu' если нужно
    amp=True
)
model.save("lenta_yolo_model.pt")
```

```python
input_path = '/kaggle/input/datasets/mashamalyshkina/lenta-yolo-dataset/train/images/25_12-20_ts0_jpg.rf.C3oOKh1XftNny50xRHin.jpg'
```

```python
img = Image.open(input_path)
#img_rotated = img.rotate(90, expand=True)
```
